# 03. Đánh giá Thực nghiệm: U-Net vs Attention U-Net
**Mục tiêu:**
- So sánh định lượng các chỉ số Dice, IoU, Precision, Recall giữa Baseline U-Net và Attention U-Net
- Phân tích trực quan khả năng phát hiện tổn thương kích thước nhỏ (Microaneurysms) nhờ Attention Gates

```python
import sys, os
sys.path.insert(0, "..")
import torch
import pandas as pd
import matplotlib.pyplot as plt

hist_p = "../ai/checkpoints/idrid_unet_training_history.csv"
if os.path.exists(hist_p):
    df_hist = pd.read_csv(hist_p)
    print(df_hist)
    plt.figure(figsize=(9, 4))
    plt.subplot(1, 2, 1)
    plt.plot(df_hist["epoch"], df_hist["train_loss"], label="Train Loss")
    plt.plot(df_hist["epoch"], df_hist["val_loss"], label="Val Loss")
    plt.title("Loss Curve")
    plt.legend()
    plt.grid(True)
    plt.subplot(1, 2, 2)
    plt.plot(df_hist["epoch"], df_hist["val_dice"], label="Val Dice", color="green")
    plt.plot(df_hist["epoch"], df_hist["val_iou"], label="Val IoU", color="purple")
    plt.title("Validation Metrics")
    plt.legend()
    plt.grid(True)
    plt.tight_layout()
    plt.show()
```
