# 02. Tiền xử lý Ảnh Y sinh & Kỹ thuật Tăng cường Dữ liệu (Augmentations)
**Nội dung:**
1. Chuẩn hóa kích thước ảnh về (512, 512)
2. Cân bằng tương phản thích ứng CLAHE trên kênh Green
3. Trích xuất vùng thị trường võng mạc (FOV mask)
4. Tăng cường hình học đồng bộ giữa ảnh và mask

```python
import sys, os
sys.path.insert(0, "..")
import cv2
import numpy as np
import matplotlib.pyplot as plt
from PIL import Image
from ai.evaluation.evaluator import extract_retinal_fov
from ai.training.augmentations import RetinalAugmentor

sample_p = "../data/IDRiD/1. Original Images/b. Testing Set/IDRiD_55.jpg"
if os.path.exists(sample_p):
    img = cv2.imread(sample_p)
    img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
    img_resized = cv2.resize(img_rgb, (512, 512))
    clahe = cv2.createCLAHE(clipLimit=2.0, tileGridSize=(8, 8))
    green_clahe = clahe.apply(img_resized[:, :, 1])
    fov, fov_pixels = extract_retinal_fov(img_resized)
    print(f"Retinal FOV pixels: {fov_pixels} px")
```

```python
if os.path.exists(sample_p):
    fig, axes = plt.subplots(1, 3, figsize=(12, 4))
    axes[0].imshow(img_resized)
    axes[0].set_title("Original Fundus (512x512)")
    axes[0].axis("off")
    axes[1].imshow(green_clahe, cmap="gray")
    axes[1].set_title("Green Channel CLAHE")
    axes[1].axis("off")
    axes[2].imshow(fov, cmap="gray")
    axes[2].set_title("Extracted Retinal FOV Mask")
    axes[2].axis("off")
    plt.tight_layout()
    plt.show()
```
