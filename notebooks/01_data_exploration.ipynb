```python
import os, sys
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from PIL import Image

# Đọc dữ liệu thống kê đã tạo từ pipeline thực tế
df_idrid = pd.read_csv("../data/idrid_statistics.csv")
print("Tổng số ảnh IDRiD:", len(df_idrid))
df_idrid.head()
```

```python
df_drive = pd.read_csv("../data/drive_statistics.csv")
print("Tổng số ảnh DRIVE:", len(df_drive))
df_drive.head()
```

```python
lesions = ["MA_present", "HE_present", "EX_present", "SE_present", "OD_present"]
counts = [df_idrid[c].sum() for c in lesions]
labels = ["Microaneurysms (MA)", "Haemorrhages (HE)", "Hard Exudates (EX)", "Soft Exudates (SE)", "Optic Disc (OD)"]
colors = ["#ef4444", "#f97316", "#eab308", "#22c55e", "#3b82f6"]

plt.figure(figsize=(8, 4.5))
plt.bar(labels, counts, color=colors)
plt.title("Tỷ lệ xuất hiện tổn thương trong bộ dữ liệu IDRiD", fontsize=12, fontweight="bold")
plt.ylabel("Số lượng ảnh")
plt.xticks(rotation=20)
plt.grid(axis="y", linestyle="--", alpha=0.7)
plt.tight_layout()
plt.show()
```
